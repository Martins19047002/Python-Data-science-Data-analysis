# Indiana COVID-19 Analysis

This notebook explores the Indiana COVID-19 sample dataset, identifies trend patterns, and summarizes the research questions and hypotheses.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

data_path = Path('..') / 'data' / 'indiana_covid19_sample.csv'
df = pd.read_csv(data_path)
df['date'] = pd.to_datetime(df['date'])
df.head()
```

## Trend Analysis

We examine daily cases and deaths over time to identify major surges and trend changes.


```python
df.plot(x='date', y='new_cases', kind='line', figsize=(12, 4), color='steelblue')
plt.title('Indiana Daily COVID-19 Cases')
plt.ylabel('Cases')
plt.tight_layout()
plt.show()
```

```python
df.plot(x='date', y='new_deaths', kind='line', figsize=(12, 4), color='crimson')
plt.title('Indiana Daily COVID-19 Deaths')
plt.ylabel('Deaths')
plt.tight_layout()
plt.show()
```

## Hypotheses

1. Indiana experienced several distinct surges over the observed period.
2. Vaccination rollout corresponded with reductions in severe outcomes after early rollout stages.
3. Testing volume and positivity trends are associated with observed case detection patterns.
4. A change in case trends may lag behind broader public health interventions and social behavior changes.


```python
rolling_cases = df['new_cases'].rolling(window=7, min_periods=1).mean()
rolling_deaths = df['new_deaths'].rolling(window=7, min_periods=1).mean()

df['rolling_cases'] = rolling_cases
df['rolling_deaths'] = rolling_deaths

df[['date', 'new_cases', 'rolling_cases', 'new_deaths', 'rolling_deaths']].head()
```

## Summary

The results show that COVID-19 trends in Indiana were not linear. Wave-like spikes, declining periods, and changing public health response patterns were all visible in the data.
