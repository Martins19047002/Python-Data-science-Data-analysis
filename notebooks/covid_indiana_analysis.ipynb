# Indiana COVID-19 Data Analysis

This notebook explores a sample Indiana COVID-19 dataset and evaluates trend patterns, hypothesis questions, and public health interpretation.


```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

data_path = Path('..') / 'data' / 'indiana_covid19_sample.csv'
df = pd.read_csv(data_path)
df['date'] = pd.to_datetime(df['date'])
df.head()
```

## Trend analysis

We inspect how daily case and death counts changed over time.


```python
df.plot(x='date', y='new_cases', kind='line', figsize=(12, 4), color='steelblue')
plt.title('Indiana Daily COVID-19 Cases')
plt.ylabel('Cases')
plt.tight_layout()
plt.show()
```

```python
df.plot(x='date', y='new_deaths', kind='line', figsize=(12, 4), color='crimson')
plt.title('Indiana Daily COVID-19 Deaths')
plt.ylabel('Deaths')
plt.tight_layout()
plt.show()
```

## Hypotheses and interpretation

1. Indiana likely experienced multiple waves across the observed period.
2. Vaccination growth may have aligned with reduced severe outcomes in the later pandemic period.
3. Testing activity and positivity rate explain some, but not all, of the case pattern.
4. Public health response, immunity, and reporting lag shaped the differences between case and death trends.


```python
df['rolling_cases'] = df['new_cases'].rolling(window=7, min_periods=1).mean()
df['rolling_deaths'] = df['new_deaths'].rolling(window=7, min_periods=1).mean()

df[['date', 'new_cases', 'rolling_cases', 'new_deaths', 'rolling_deaths']].head()
```

## Conclusion

The results show that COVID-19 outbreaks were not static. Indiana saw time-varying case patterns, public health adjustments, and changes in testing and vaccination dynamics. These elements are important in any data-driven analysis of the pandemic.
